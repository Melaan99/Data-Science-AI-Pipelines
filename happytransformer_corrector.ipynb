```python
from happytransformer import HappyTextToText
text = HappyTextToText("T5", "prithivida/grammar_error_correcter_v1")
```

```python
from happytransformer import TTSettings
top_k_sampling_settings = TTSettings(do_sample=True, top_k=50, temperature=0.7, min_length=1, max_length=50)
```

```python
result = text.generate_text("I likes to eatt. applees", args=top_k_sampling_settings)
print(result.text)
```

```python
result_2 = text.generate_text("After we played ches, I read my fav bok and wen to bed. ", args=top_k_sampling_settings)
print(result_2.text)
```
